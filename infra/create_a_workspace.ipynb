```python
# Establish connection to workspace

from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential

ml_client = MLClient.from_config(
    credential=DefaultAzureCredential()
)

print(ml_client)
```

```python
# Creating a unique workspace name with current datetime to avoid conflicts
from azure.ai.ml.entities import Workspace
import datetime

basic_workspace_name = "mlw-dev" + datetime.datetime.now().strftime(
    "%Y%m%d%H%M"
)

ws_basic = Workspace(
    name=basic_workspace_name,
    location="westus",
    display_name="Basic workspace-example",
    description="This is dev workspace for NER MLOps demo",
    hbi_workspace=False,
    tags=dict(purpose="dev"),
)

ws_basic = ml_client.workspaces.begin_create(ws_basic).result()
print(ws_basic)
```
