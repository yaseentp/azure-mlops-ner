```python
# Establish connection to workspace

from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential

ml_client = MLClient.from_config(
    credential=DefaultAzureCredential()
)

print(ml_client)
```

```python
pipeline_job_name = "dynamic_grape_rh2tfs1yrt"
pipeline_job = ml_client.jobs.get(
    name=pipeline_job_name
)

print("Status:", pipeline_job.status)
print("Outputs:", pipeline_job.outputs)
```

```python

# download model; output of the pipeline run

ml_client.jobs.download(
    name=pipeline_job_name,
    output_name="trained_model",
    download_path="./downloaded_model",
    
)
```

```python
from transformers import (
    AutoTokenizer,
    AutoModelForTokenClassification,
)

model_path = "./downloaded_model/named-outputs/trained_model"

tokenizer = AutoTokenizer.from_pretrained(model_path)

model = AutoModelForTokenClassification.from_pretrained(model_path)

print("Model loaded successfully!")
```

```python
from transformers import pipeline

model_path = "./downloaded_model/named-outputs/trained_model"

ner_pipeline = pipeline(
    task="token-classification",
    model=model_path,
    tokenizer=model_path,
    aggregation_strategy="simple",
)
```

```python
text = """
Patient admitted on 12/05/2024.
Blood pressure is 120/80.
Temperature is 38 degrees.
Discharged on 15/05/2024.
"""

results = ner_pipeline(text)

for entity in results:
    print(entity)
```
